# Faithful QDENN fixed-split in-domain runner

This notebook runs one QDENN dataset unit with training seeds 101, 202, and 303. Split seed 101 is fixed inside the audited runner. Checkpoints and token caches are stored on Google Drive, so rerunning this notebook resumes interrupted epochs.

```python
from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
# Edit only these three values. Keep the accelerator set to CPU.
REPO_ROOT = Path('/content/drive/MyDrive/Python-VulCode-Detection')
MODEL = 'qdenn'
DATASET = 'benchmarkpython'  # benchmarkpython, vudenc, realvuln_human

RUNNER = REPO_ROOT / 'QDENN_SOTA/experiments/faithful/run_in_domain.py'
OUTPUT = Path('/content/drive/MyDrive/quantum_sota_fixedsplit101/results')
CACHE = Path('/content/drive/MyDrive/quantum_sota_fixedsplit101/cache')
assert RUNNER.is_file(), f'Missing runner: {RUNNER}'
```

```python
import subprocess, sys
command = [
    sys.executable, str(RUNNER),
    '--model', MODEL,
    '--dataset', DATASET,
    '--epochs', '10',
    '--output', str(OUTPUT),
    '--cache', str(CACHE),
]
print(' '.join(command), flush=True)
subprocess.run(command, check=True)
```
